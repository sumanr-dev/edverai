mod3:Prompt Engineering: Event & Travel Assistant

### Scenario
We use an **Event & Travel Assistant** that helps users understand tickets, bookings, venues, timings, and travel information.

Each concept includes a simple explanation and a small example.


## 1. What is Prompt Engineering?

Prompt engineering means writing clear instructions for an AI model.
A good prompt tells the model what to do, what information to use, and how the answer should look.
For our scenario, prompts help the assistant answer event and travel questions consistently.

**Concept / Method used:** Prompt Engineering

```python
prompt = """
You are an event travel assistant.
Answer the user's question using the provided ticket information.
Keep the answer short and easy to understand.
"""

print(prompt)
```

## 2. Zero-Shot Prompting

Zero-shot prompting asks the model to perform a task without giving examples.
The model receives only the instruction and the user's request.
It is useful for simple and familiar tasks.

**Concept / Method used:** Zero-Shot Prompting

```python
prompt = """
You are an event assistant.
Explain why a user should keep their booking reference with them.
Answer in two simple sentences.
"""

print(prompt)
```

## 3. Few-Shot Prompting

Few-shot prompting gives the model a few examples before the actual request.
The examples show the model the expected style or format.
This is useful when we want consistent responses.

**Concept / Method used:** Few-Shot Prompting / In-Context Learning

```python
prompt = """
Convert questions into polite assistant responses.

Example:
User: Where is the venue?
Assistant: The venue is shown on your booking.

Example:
User: What time does it start?
Assistant: The start time is shown on your ticket.

Now answer:
User: Which gate should I use?
"""

print(prompt)
```

## 4. Instructional Prompting

Instructional prompting gives clear rules for completing a task.
We can control length, language, format, and what the model should avoid.
This makes the output more predictable.

**Concept / Method used:** Instructional Prompting / Output Constraints

```python
prompt = """
You are an event assistant.

Task:
Create an arrival checklist.

Rules:
- Give exactly 4 points.
- Use simple English.
- Keep each point under 12 words.
- Do not add unrelated information.
"""

print(prompt)
```

## 5. Reasoning Prompting

Reasoning prompts help the model handle tasks that need multiple steps.
For example, the assistant may calculate an arrival time from an event start time.
The prompt can ask the model to work through the task and provide the useful result.

**Concept / Method used:** Reasoning Prompting / Task Decomposition

```python
prompt = """
An event starts at 7:30 PM.
The user wants to arrive 45 minutes early.

Work out the recommended arrival time.
Give the calculation briefly and then the final answer.
"""

print(prompt)
```

## 6. Role-Based Prompting

Role-based prompting gives the AI a specific role or personality.
Here, the model acts as a friendly event and travel concierge.
The role helps control the tone and type of response.

**Concept / Method used:** Role Prompting / Persona

```python
prompt = """
You are a friendly event concierge.
Speak clearly and politely.
Help users understand their ticket and travel information.
Avoid technical language.
"""

print(prompt)
```

## 7. Context-Based Prompting

Context-based prompting supplies useful information along with the question.
The model can use the ticket details as context before answering.
This is especially useful when the answer must be based on known information.

**Concept / Method used:** Context Injection / Grounded Prompting

```python
ticket_context = {
    "event": "Music Festival",
    "date": "Saturday",
    "time": "7:30 PM",
    "venue": "City Arena",
    "gate": "Gate B",
    "seat": "A24"
}

prompt = f"""
Use only the following ticket information:

{ticket_context}

User question:
What gate should I enter through?

Give a short answer.
"""

print(prompt)
```

## 8. Grounded Prompting

Grounded prompting tells the model to stay connected to trusted information.
It should not guess details that are missing from the ticket.
If the information is unavailable, the model should clearly say so.

**Concept / Method used:** Grounding / Hallucination Control

```python
prompt = """
Answer only from the booking information provided below.

Rules:
- Do not guess.
- Do not invent missing details.
- If the answer is not available, say:
  "That information is not available in the booking."

Booking:
Venue: City Arena
Gate: Gate B
"""

print(prompt)
```

## 9. Constraint Prompting

Constraint prompting places limits on the model's response.
We can specify the number of sentences, words, bullets, or allowed information.
This is useful for chatbots and voice assistants where short answers are preferred.

**Concept / Method used:** Constraints / Output Control

```python
prompt = """
Answer the ticket question using these constraints:
- Maximum 2 sentences.
- Maximum 30 words.
- Simple English.
- No markdown.
- Do not repeat the question.
"""

print(prompt)
```

## 10. Format Prompting

Format prompting tells the model exactly how the answer should be structured.
For example, we can request JSON fields for event information.
A fixed format makes the output easier for another program to process.

**Concept / Method used:** Structured Output / Format Control

```python
prompt = """
Extract the event information.

Return only this JSON structure:

{
  "event": "",
  "date": "",
  "time": "",
  "venue": "",
  "gate": ""
}

Do not add explanations.
"""

print(prompt)
```

## 11. RTCF Prompting

RTCF is a simple framework for building clear production prompts.
R means Role, T means Task, C means Context, and F means Format.
Putting these parts together makes the prompt easier to understand and maintain.

**Concept / Method used:** RTCF – Role, Task, Context, Format

```python
prompt = """
ROLE:
You are a helpful event travel assistant.

TASK:
Answer the user's question about their ticket.

CONTEXT:
The user's ticket says the event starts at 7:30 PM at City Arena.

FORMAT:
Give one short sentence.
Do not guess missing information.
"""

print(prompt)
```

## 12. Prompt Chaining

Prompt chaining breaks one large task into smaller prompt steps.
For example, one prompt can extract ticket details and another can answer the user's question.
This makes each stage easier to test and control.

**Concept / Method used:** Prompt Chaining / Multi-Step Prompting

```python
extract_prompt = """
Extract the event name, date, time, venue, gate, and seat
from the supplied ticket information.
"""

answer_prompt = """
Using the extracted ticket details, answer the user's question
in one short sentence.
"""

print("Step 1 – Extraction:")
print(extract_prompt)

print("\nStep 2 – Answer:")
print(answer_prompt)
```

## 13. Comparative Prompting

Comparative prompting asks the model to compare two or more pieces of information.
For example, we can compare two travel options using the same criteria.
The prompt should clearly state what attributes should be compared.

**Concept / Method used:** Comparative Prompting / Criteria-Based Analysis

```python
prompt = """
Compare these two travel options:

Option A: Train, 3 hours, ₹900
Option B: Bus, 4.5 hours, ₹600

Compare:
- Travel time
- Price

Give a simple table.
Do not add other criteria.
"""

print(prompt)
```

## 14. Multi-Model Prompt Testing

The same prompt can be tested with different language models.
This helps us observe differences in response quality, speed, and instruction following.
The prompt itself stays the same so the comparison is meaningful.

**Concept / Method used:** Model Comparison / Prompt Benchmarking

```python
prompt = """
Explain what a booking reference is.
Use simple English.
Answer in exactly two sentences.
"""

models = ["Model A", "Model B", "Model C"]

for model in models:
    print(f"{model}: test with the same prompt")
    print(prompt)
    print()
```

## 15. Prompt Template

A prompt template keeps the main instructions fixed while allowing the input to change.
This is useful when the assistant receives many different tickets or questions.
Only the variables, such as the question and booking details, need to change.

**Concept / Method used:** Prompt Template / Variable Injection

```python
booking = "Event: Music Festival | Time: 7:30 PM | Venue: City Arena"
question = "What time does the event start?"

prompt = f"""
You are an event assistant.

Booking:
{booking}

User question:
{question}

Answer in one short sentence.
"""

print(prompt)
```

## 16. Prompt Safety and Guardrails

A production prompt should clearly define what the model must not do.
For a booking assistant, it should not invent ticket details or claim information that is not available.
These rules reduce unwanted or misleading responses.

**Concept / Method used:** Prompt Guardrails / Safety Constraints

```python
prompt = """
You are a ticket information assistant.

IMPORTANT RULES:
1. Use only supplied ticket information.
2. Never invent a gate, seat, date, or time.
3. If information is missing, say it is unavailable.
4. Do not reveal private information that is not needed.
5. Keep the answer short and clear.
"""

print(prompt)
```

## 17. Final Prompt – Event & Travel Assistant

This final prompt combines the main techniques into one practical prompt.
It uses a role, task, context, constraints, grounding, and output format.
This is the type of prompt we can use as the foundation of the complete assistant.

**Concept / Method used:** RTCF + Grounding + Constraints + Role-Based Prompting

```python
ticket = {
    "event": "Music Festival",
    "date": "Saturday",
    "time": "7:30 PM",
    "venue": "City Arena",
    "gate": "Gate B",
    "seat": "A24"
}

user_question = "What time does the event start and which gate should I use?"

final_prompt = f"""
ROLE:
You are a friendly event and travel assistant.

TASK:
Answer the user's question using the ticket information.

CONTEXT:
{ticket}

USER QUESTION:
{user_question}

RULES:
- Use only the provided ticket information.
- Do not guess or invent details.
- If information is missing, say it is unavailable.
- Keep the answer short.
- Use simple English.
- Do not use markdown.

FORMAT:
Return only the final answer.
"""

print(final_prompt)
```

## 18. Prompt Concepts Summary

These prompt types can be combined depending on the task.
Simple questions may need zero-shot prompting, while complex workflows can use few-shot, RTCF, chaining, and guardrails.
The main goal is to make the model's instructions clear, controlled, and grounded.

**Concept / Method used:** Prompt Engineering Summary
