## Step 2. Define the Student Dataset

The project stores each student's name and marks for three subjects. A dictionary is suitable because each student name can be associated with a collection of subject marks.

The values are integers because the marks represent numerical measurements. This structure will later be passed to reusable functions for analysis.

```python
students = {
    "Asha": {"Python": 82, "Database": 76, "Statistics": 88},
    "Rahul": {"Python": 58, "Database": 64, "Statistics": 55},
    "Priya": {"Python": 94, "Database": 91, "Statistics": 89},
    "Arun": {"Python": 72, "Database": 68, "Statistics": 74},
    "Neha": {"Python": 43, "Database": 49, "Statistics": 46}
}

print("Number of students:", len(students))
```

The important concept in this cell is structured data. Each student is represented by a nested dictionary containing the marks for three subjects. This allows the program to access individual values through dictionary keys.

## Step 3. Define the Academic Result Function

The result classification is separated into its own function. This follows the reusable function approach from the reference notebook.

The function receives an average as a parameter and returns a result category. It does not collect input or print a report because its single responsibility is to classify performance.

```python
def check_result(average):
    if average >= 75:
        return "Distinction"
    elif average >= 50:
        return "Pass"
    else:
        return "Fail"
```

The function demonstrates parameters, conditional statements, comparison operators, and return values.

The conditions are evaluated from top to bottom. Once a condition is true, Python returns the corresponding result and leaves the function.

## Step 4. Calculate a Student Average

This function receives one student's subject marks and calculates the average.

The function uses the dictionary values rather than depending on specific subject names. This makes the function reusable for any student with the same data structure.

```python
def calculate_average(subject_marks):
    total_marks = sum(subject_marks.values())
    subject_count = len(subject_marks)
    return total_marks / subject_count
```

The parameter subject_marks represents the actual dictionary supplied when the function is called.

The function returns a numerical value instead of printing it. Returning values makes the function useful to other parts of the program.

## Step 5. Build a Student Performance Profile

The next function combines the average calculation and result classification functions.

This demonstrates how smaller functions can be connected to create a larger application workflow. Each function remains responsible for one clear operation.

```python
def create_student_profile(name, subject_marks):
    average = calculate_average(subject_marks)
    result = check_result(average)

    return {
        "name": name,
        "marks": subject_marks,
        "average": average,
        "result": result
    }
```

The function accepts two parameters. The name identifies the student and subject_marks contains the academic information.

The function returns a new dictionary containing the original information together with the calculated average and result. This creates a reusable student profile that can be processed by later functions.

## Step 6. Generate Profiles for All Students

The previous function works for one student. This cell applies the same reusable logic to every student in the dataset.

The result is stored in a new dictionary called profiles. The original student dataset remains unchanged.

```python
def build_all_profiles(students):
    profiles = {}

    for name, subject_marks in students.items():
        profiles[name] = create_student_profile(name, subject_marks)

    return profiles


profiles = build_all_profiles(students)

print("Profiles created:", len(profiles))
```

The expected output is Profiles created: 5.

This cell demonstrates function reuse and iteration. Instead of duplicating the calculation logic for every student, the program calls the same function repeatedly.

## Step 7. Calculate Class Statistics

A useful academic report should contain more than individual results. The following function calculates the class average and counts the number of students in each result category.

This function demonstrates how multiple values can be processed together and returned as a structured dictionary.

```python
def calculate_class_statistics(profiles):
    averages = []
    distinction_count = 0
    pass_count = 0
    fail_count = 0

    for profile in profiles.values():
        averages.append(profile["average"])

        if profile["result"] == "Distinction":
            distinction_count += 1
        elif profile["result"] == "Pass":
            pass_count += 1
        else:
            fail_count += 1

    class_average = sum(averages) / len(averages)

    return {
        "class_average": class_average,
        "distinction_count": distinction_count,
        "pass_count": pass_count,
        "fail_count": fail_count
    }
```

The function keeps the statistical logic separate from the display logic. This makes the calculation easier to test and maintain.

For the supplied dataset, the class average is approximately 68.47. There are two distinction results, two pass results, and one fail result.

## Step 8. Identify the Highest Performing Student

The next reusable function identifies the student with the highest average.

The function receives the profiles dictionary and returns the complete profile of the highest performing student.

```python
def find_top_student(profiles):
    return max(profiles.values(), key=lambda profile: profile["average"])
```

The max function compares the average value from each profile. The function returns the profile rather than only the name so that the report can display additional information when required.

For the supplied dataset, the highest average belongs to Priya.

## Step 9. Define Campus Information

The API requires geographic coordinates to retrieve weather information. The following variables define the campus location used by this project.

Keeping these values in variables makes the API function reusable for another campus without changing the function itself.

```python
campus_name = "Bhopal Campus"
latitude = 23.2599
longitude = 77.4126

print("Campus:", campus_name)
print("Latitude:", latitude)
print("Longitude:", longitude)
```

The variables latitude and longitude are floating point values because geographic coordinates can contain decimal values.

The values are then passed as arguments to the API function.

## Step 10. Create a Reusable Weather API Function

This function applies the API workflow from the reference API integration notebook.

The workflow is request, status check, JSON conversion, and data extraction.

The function returns a dictionary when the request succeeds and returns None when the request fails.

```python
def get_campus_weather(latitude, longitude):
    url = "https://api.open-meteo.com/v1/forecast"

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "current": "temperature_2m,relative_humidity_2m,weather_code",
        "timezone": "auto"
    }

    response = requests.get(url, params=params, timeout=10)

    if response.status_code == 200:
        return response.json()

    return None
```

The URL identifies the API endpoint. The params dictionary contains the requested geographic coordinates and current weather fields.

The status code is checked before the JSON response is processed. This is important because an unsuccessful response should not be treated as valid data.

The timeout prevents the application from waiting indefinitely for a response.

## Step 11. Retrieve Weather Data

The API function is now called using the campus coordinates.

The returned JSON object is stored in weather_data. This demonstrates the separation between an API function and the application logic that uses its result.

```python
weather_data = get_campus_weather(latitude, longitude)

if weather_data is not None:
    print("Weather data retrieved successfully")
else:
    print("Weather data could not be retrieved")
```

When the API is available, the expected output is Weather data retrieved successfully.

The actual weather values can change because they are retrieved from a live external service.

## Step 12. Extract Nested JSON Values

API responses commonly contain nested dictionaries. The following function extracts only the fields required by the final report.

This follows the same principle demonstrated in the reference API notebook where specific values are extracted from a larger JSON response.

```python
def extract_weather_summary(weather_data):
    if weather_data is None:
        return None

    current = weather_data.get("current", {})

    return {
        "temperature": current.get("temperature_2m"),
        "humidity": current.get("relative_humidity_2m"),
        "weather_code": current.get("weather_code"),
        "time": current.get("time")
    }
```

The API response contains a current object inside the main JSON dictionary. The function first accesses that nested object and then extracts the required values.

The get method is used so that missing fields do not immediately produce a KeyError. This provides a safer way to process external data.

## Step 13. Convert Weather Codes into Descriptions

The weather API provides a numerical weather code. A separate function converts selected codes into readable descriptions.

Keeping this conversion in a function makes the application easier to extend when additional weather codes are required.

```python
def describe_weather_code(code):
    descriptions = {
        0: "Clear sky",
        1: "Mainly clear",
        2: "Partly cloudy",
        3: "Overcast",
        45: "Fog",
        48: "Depositing rime fog",
        51: "Light drizzle",
        61: "Light rain",
        63: "Moderate rain",
        65: "Heavy rain",
        80: "Rain showers",
        95: "Thunderstorm"
    }

    return descriptions.get(code, "Weather condition not classified")
```

The descriptions dictionary maps API codes to readable text. The get method provides a default message when a code is not included in the mapping.

This is another example of separating processing logic from display logic.

## Step 14. Prepare the Complete Report Data

The following function brings together the academic statistics, top student information, campus information, and weather information.

The function prepares data for reporting but does not print the final report. This keeps the application modular.

```python
def prepare_report(profiles, campus_name, weather_summary):
    statistics = calculate_class_statistics(profiles)
    top_student = find_top_student(profiles)

    report = {
        "generated_at": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "campus": campus_name,
        "class_average": statistics["class_average"],
        "distinction_count": statistics["distinction_count"],
        "pass_count": statistics["pass_count"],
        "fail_count": statistics["fail_count"],
        "top_student": top_student,
        "weather": weather_summary
    }

    return report
```

The function demonstrates how multiple reusable functions can be combined into a single application workflow.

The final report is represented as a dictionary. This makes the report data easy to inspect, reuse, or export later.

## Step 15. Display an Individual Student Profile

The display function is responsible only for formatting output. This follows the single responsibility principle used throughout the reusable functions notebook.

```python
def display_student_profile(profile):
    print("Student Performance Profile")
    print("Student:", profile["name"])

    for subject, marks in profile["marks"].items():
        print(f"{subject}: {marks}")

    print(f"Average: {profile['average']:.2f}")
    print("Result:", profile["result"])
```

The function receives a complete profile as an argument and formats its contents for display.

The function does not calculate the average or determine the result. Those responsibilities have already been assigned to separate functions.

## Step 17. Prepare the Weather Summary

The raw API response is now converted into a smaller weather summary that contains only the values needed by the report.

```python
weather_summary = extract_weather_summary(weather_data)

if weather_summary is not None:
    weather_summary["description"] = describe_weather_code(
        weather_summary["weather_code"]
    )

print(weather_summary)
```

The output will depend on the live API response. The important fields are temperature, humidity, weather code, time, and the human readable weather description.

## Step 18. Build the Complete Application Report

The final report object is created by combining all previously prepared information.

This is the point where the separate modules of the project become one integrated application.

```python
report = prepare_report(
    profiles,
    campus_name,
    weather_summary
)

print("Report prepared successfully")
```

The expected output is Report prepared successfully.

At this stage the application has calculated academic results, class statistics, the highest performing student, and the latest available campus weather information.

## Step 19. Display the Final Professional Report

The final display function presents the prepared data in a structured format.

The function does not perform calculations or API calls. It only displays information that has already been processed.

```python
def display_final_report(report):
    print("STUDENT PERFORMANCE AND CAMPUS READINESS REPORT")
    print()
    print("Campus:", report["campus"])
    print("Generated at:", report["generated_at"])
    print()
    print("Academic Summary")
    print("Class average:", f"{report['class_average']:.2f}")
    print("Distinction students:", report["distinction_count"])
    print("Pass students:", report["pass_count"])
    print("Fail students:", report["fail_count"])
    print()
    print("Highest Performing Student")
    print("Name:", report["top_student"]["name"])
    print("Average:", f"{report['top_student']['average']:.2f}")
    print("Result:", report["top_student"]["result"])
    print()
    print("Campus Weather")

    if report["weather"] is not None:
        print("Temperature:", report["weather"]["temperature"])
        print("Humidity:", report["weather"]["humidity"])
        print("Condition:", report["weather"]["description"])
        print("Observed at:", report["weather"]["time"])
    else:
        print("Weather information is currently unavailable")
```

The report separates academic information from campus information so that the output is easy to understand.

The weather section also handles an unavailable API response without stopping the complete academic report. This demonstrates basic failure handling for external services.

## Step 20. Run the Complete Application

The final cell calls the display function. All calculations and API processing have already been placed inside reusable functions.

The main execution section is therefore short and easy to understand.

```python
display_final_report(report)
```
